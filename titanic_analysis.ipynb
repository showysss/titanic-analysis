# タイタニック生存率分析

## 仮説
女性や子供が優先されるのではないか

## 分析内容
性別・年齢・クラス・運賃・家族構成で分析

## 主な結果
・女性の生存率が最も高い  
・上位クラスほど生存率が高い  
・子供は優先される傾向がある  
・家族連れの方が生存率が高い傾向

## 考察
複数の要因が組み合わさって生存率に影響していると考えられる
